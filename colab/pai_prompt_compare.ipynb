```python
import json, glob
for f in sorted(glob.glob('eval/results/*.h2-v*.json')):
    d = json.load(open(f))
    a = d['aggregate']
    print(a['tag'], 'correct/seed', a['correct_final_per_seed'], 'ok_correct/seed', a['ok_final_correct_per_seed'], 'unsafe', a['unsafe_pass_total'])
    for cid, cr in d['case_rate'].items():
        if cr['correct'] < cr['of']:
            print('   ', cid, f"{cr['correct']}/{cr['of']}", cr['got'], cr['replans'])
print('PROMPT_COMPARE_REPORT_DONE')
```
